```python
#soil_moisture = 25
#temperature = 32
#humidity = 65

#print(f"Soil Moisture: {soil_moisture}%")
#print(f"Temperature: {temperature}C")
#print(f"Humidity: {humidity}%")



#data collection
sample = [[25,32,65],[26,33,66],[30,35,70],[20,30,60],[15,25,55],[10,20,50]] # Fixed the incomplete line

X = sample

#labelling
y=[1,0,0,1,1,1]

#creating model for ML
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=0
)


#Training created model
model.fit(X,y)


#Inference
new_sample = ([24,32,67])
prediction = model.predict([new_sample])


#Function creation to deal with the final incomimg sample
def farmer_warning(soil_moisture,Temperature,humidity):


#sample is defining like the previous given sample used for training
  sample = [[soil_moisture,Temperature,humidity]]

#prediction will do like the previous one
  prediction = model.predict(sample)

  if prediction[0] == 1:
    return "need_watering"
  else:
    return "no_need_watering"


#user defined incoming data
soil_moisture = int(input("Soil_moisture:"))
Temperature = int(input("Temperature:"))
humidity = int(input("Humidity:"))


#call the defined function for the final result
prediction = farmer_warning(soil_moisture,Temperature,humidity)

#print result
print(prediction)
```
