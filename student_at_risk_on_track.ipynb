```python
#study_hour = 10
#attendance = 60
#assignemnt = 85
#print(study_hour,"hour",attendance,"percentage",assignemnt,"percentage")

#sample = [study_hour,attendance,assignemnt]
#print(sample)

sample = [[10,60,85],[2,45,67],[1,50,70],[4,55,75],[5,60,77]]
#print(sample[4])

X=sample
#print(X)

y = [0,1,1,0,0]

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=3,
    random_state=0
)

model.fit(X, y)

new_student = [[4, 72, 70]]

prediction = model.predict(new_student)

def student_warning(study_hours, attendance, assignments):

    sample = [[study_hours, attendance, assignments]]

    prediction = model.predict(sample)

    if prediction[0] == 1:
        return "AT RISK"
    else:
        return "ON TRACK"
study_hours = int(input("Enter the number of study hours: "))
attendance = int(input("Enter the attendance percentage: "))
assignments = int(input("Enter the assignment scores: "))

result = student_warning(study_hours,attendance,assignments)

print("System output:", result)
```

চলুন দেখি কেমন ডেটা 'AT RISK' হিসাবে চিহ্নিত হয়।

```python
# To see an 'AT RISK' example, we can use values from the training data that correspond to an 'AT RISK' label (y=1).
# For example, student data [2, 45, 67] was labeled as 'AT RISK' (index 1 of y list).
result_at_risk = student_warning(2, 45, 67)

print("System output for an 'AT RISK' student:", result_at_risk)
```

```python
import random

def generate_at_risk_data():
    # Generate random values within 'AT RISK' ranges observed in training data
    study_hours = random.randint(0, 3) # Low study hours
    attendance = random.randint(40, 50) # Low attendance
    assignments = random.randint(60, 70) # Low assignment scores
    return [study_hours, attendance, assignments]

# Generate random 'AT RISK' data
random_at_risk_student_data = generate_at_risk_data()
print(f"Randomly generated student data: {random_at_risk_student_data}")

# Predict using the student_warning function
result_random = student_warning(random_at_risk_student_data[0], random_at_risk_student_data[1], random_at_risk_student_data[2])

print(f"System output for randomly generated data: {result_random}")
```
