```python
#study_hour = 10
#attendance = 60
#assignemnt = 85
#print(study_hour,"hour",attendance,"percentage",assignemnt,"percentage")

#sample = [study_hour,attendance,assignemnt]
#print(sample)

sample = [[10,60,85],[2,45,67],[1,50,70],[4,55,75],[5,60,77]]
#print(sample[4])

X=sample
#print(X)

y = [0,1,1,0,0]

from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=3,
    random_state=0
)

model.fit(X, y)

new_student = [[4, 72, 70]]

prediction = model.predict(new_student)

def student_warning(study_hours, attendance, assignments):

    sample = [[study_hours, attendance, assignments]]

    prediction = model.predict(sample)

    if prediction[0] == 1:
        return "AT RISK"
    else:
        return "ON TRACK"
study_hours = int(input("Enter the number of study hours: "))
attendance = int(input("Enter the attendance percentage: "))
assignments = int(input("Enter the assignment scores: "))

result = student_warning(study_hours,attendance,assignments)

print("System output:", result)
```
